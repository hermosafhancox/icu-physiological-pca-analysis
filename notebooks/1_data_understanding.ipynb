# Investigating First 5 Patients

```python
import glob

files = glob.glob("../data/raw/training/training_setA/*.psv") # find every file ending in .psv 

len(files)
```

```python
sample_files = files[:5]
sample_files
```

```python
sample_patients = [pd.read_csv(file, sep="|") for file in sample_files] # reads psv file into pandas dataframe and loops through files 1-5
len(sample_patients)
```

```python
[len(patient) for patient in sample_patients] # number of observations for each patient
```

```python
[patient.shape[1] for patient in sample_patients] # checks whether same 41 columns exist for every patient
```

```python
missing_by_patient = [patient.isna().mean().mean() * 100 for patient in sample_patients] # overall percentage of missing measurements for each patient
missing_by_patient
```

```python
variable_stats = [patient.describe() for patient in sample_patients] # calculate basic statistical properties of variables across 5 patients
variable_stats
```

```python
# function to find percentage of missing values per variable across 5 patients
missing_by_variable = pd.concat(
    [patient.isna().mean() * 100 for patient in sample_patients],
    axis = 1
)

missing_by_variable.columns = [f"Patient {i+1}" for i in range(len(sample_patients))]

missing_by_variable["Mean Missing %"] = missing_by_variable.mean(axis=1)

missing_by_variable.sort_values("Mean Missing %", ascending=False)
```
